## Задача 4. Аналіз соціальної мережі компанії

**Умова:**

У невеликій IT-компанії працює 6 людей. HR-відділ хоче проаналізувати комунікацію в корпоративному месенджері: хто з ким спілкується. Вам надали дані про зв'язки між співробітниками.

**Співробітники та їхні контакти (неорієнтований граф):**

* **Анна** спілкується з: Богдан, Віктор, Ганна
* **Богдан** спілкується з: Анна, Віктор, Дмитро
* **Віктор** спілкується з: Анна, Богдан, Ганна, Дмитро
* **Ганна** спілкується з: Анна, Віктор, Євген
* **Дмитро** спілкується з: Богдан, Віктор, Євген
* **Євген** спілкується з: Ганна, Дмитро

**Напишіть програму на Python, яка виконує наступні кроки:**

1.  Побудуйте **три різні представлення** цього графа:
    * Матриця суміжності (як вкладені списки).
    * Список суміжності (як словник).
    * Список ребер (як список кортежів).
2.  Обчисліть **степінь кожної вершини** (кількість зв'язків кожного співробітника). Визначте, хто найбільш комунікабельний, а хто — найменш.
3.  **Перевірте теорему про суму степенів:** сума степенів усіх вершин дорівнює подвоєній кількості ребер. Виведіть суму всіх степенів, кількість ребер у графі та підтвердження, що рівність виконується.

```python
import pandas as pd
from IPython.display import display

names = ['Анна', 'Богдан', 'Віктор', 'Ганна', 'Дмитро', 'Євген']
n = len(names)

adj_list = {
    'Анна': ['Богдан', 'Віктор', 'Ганна'],
    'Богдан': ['Анна', 'Віктор', 'Дмитро'],
    'Віктор': ['Анна', 'Богдан', 'Ганна', 'Дмитро'],
    'Ганна': ['Анна', 'Віктор', 'Євген'],
    'Дмитро': ['Богдан', 'Віктор', 'Євген'],
    'Євген': ['Ганна', 'Дмитро']
}

adj_matrix = [[0] * n for _ in range(n)]
name_to_idx = {name: i for i, name in enumerate(names)}

for person, neighbors in adj_list.items():
    row_idx = name_to_idx[person]
    for neighbor in neighbors:
        col_idx = name_to_idx[neighbor]
        adj_matrix[row_idx][col_idx] = 1

edges = []
seen_edges = set()
for person, neighbors in adj_list.items():
    for neighbor in neighbors:
        edge = tuple(sorted((person, neighbor)))
        if edge not in seen_edges:
            edges.append(edge)
            seen_edges.add(edge)

degrees = {person: len(neighbors) for person, neighbors in adj_list.items()}
max_degree_val = max(degrees.values())
min_degree_val = min(degrees.values())

most_communicative = [k for k, v in degrees.items() if v == max_degree_val]
least_communicative = [k for k, v in degrees.items() if v == min_degree_val]

sum_degrees = sum(degrees.values())
num_edges = len(edges)
theorem_check = sum_degrees == 2 * num_edges

print("Красивий Вивід:")
print("-" * 30)
print("1. Матриця суміжності:")
df_matrix = pd.DataFrame(adj_matrix, index=names, columns=names)
display(df_matrix)

print("\n2. Аналіз активності (Степені вершин):")
df_degrees = pd.DataFrame(list(degrees.items()), columns=['Співробітник', 'Кількість зв\'язків'])
df_degrees = df_degrees.sort_values(by='Кількість зв\'язків', ascending=False)
display(df_degrees)

print(f"\nНайбільш комунікабельні ({max_degree_val} зв'язки): {', '.join(most_communicative)}")
print(f"Найменш комунікабельні ({min_degree_val} зв'язки): {', '.join(least_communicative)}")

print("\n3. Перевірка теореми про суму степенів:")
print(f"   Сума степенів: {sum_degrees}")
print(f"   Кількість ребер: {num_edges} (Подвоєна: {2 * num_edges})")
print(f"   Результат перевірки: {'Рівність виконується' if theorem_check else '❌ Помилка'}")

print("\n" + "="*50 + "\n")

print("Технічний вивід (Raw Data):")
print(f"Список суміжності (adj_list) = {adj_list}")
print(f"\nМатриця суміжності (adj_matrix) = {adj_matrix}")
print(f"\nСписок ребер ({len(edges)} унікальних) = {edges}")
```

### **Аналітичні обчислення**

#### 1. Візуалізація зв'язків (ручний підрахунок ребер)

Випишемо всі унікальні пари (ребра графа $E$), щоб не рахувати двічі (наприклад, "Анна-Богдан" і "Богдан-Анна" — це одне ребро):

1.  Анна — Богдан
2.  Анна — Віктор
3.  Анна — Ганна
4.  Богдан — Віктор
5.  Богдан — Дмитро
6.  Віктор — Ганна
7.  Віктор — Дмитро
8.  Ганна — Євген
9.  Дмитро — Євген

**Всього ребер:** $|E| = 9$.

#### 2. Обчислення степенів вершин ($deg(v)$)

Степінь вершини — це кількість ліній, що виходять з неї:

* $deg(\text{Анна}) = 3$ (Богдан, Віктор, Ганна)
* $deg(\text{Богдан}) = 3$ (Анна, Віктор, Дмитро)
* $deg(\text{Віктор}) = 4$ (Анна, Богдан, Ганна, Дмитро)
* $deg(\text{Ганна}) = 3$ (Анна, Віктор, Євген)
* $deg(\text{Дмитро}) = 3$ (Богдан, Віктор, Євген)
* $deg(\text{Євген}) = 2$ (Ганна, Дмитро)

#### 3. Перевірка "Леми про рукостискання" (Handshaking Lemma)

Теорема стверджує, що сума степенів усіх вершин графа дорівнює подвоєній кількості ребер:

$$\sum_{v \in V} deg(v) = 2 \cdot |E|$$

**Ліва частина (Сума степенів):**

$$3 + 3 + 4 + 3 + 3 + 2 = 18$$

**Права частина (Подвоєні ребра):**

$$2 \cdot 9 = 18$$

$$18 = 18$$

**Висновок:** Теорема виконується.
